```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Re-create df_raw and df with initial categorical types
data = {
    'size': ['XL','L','M','L','M'],
    'color': ['red','green','blue','green','red'],
    'gender': ['female','male','male','female','female'],
    'price': [199.0,89.0,99.0,129.0,79.0],
    'weight': [500,450,300,380,410],
    'bought': ['yes','no','yes','no','yes']
}
df_raw = pd.DataFrame(data=data)
df = df_raw.copy()

for col in ['size','color','gender','bought']:
    df[col] = df[col].astype('category')
df['weight'] = df['weight'].astype('float')

# Apply LabelEncoder for 'bought'
le = LabelEncoder()
df['bought'] = le.fit_transform(df['bought'])

# Apply StandardScaler for 'price' and 'weight'
scaler = StandardScaler()
df[['price','weight']] = scaler.fit_transform(df[['price', 'weight']])

# Apply One-Hot Encoding with numerical output for dummy variables
df = pd.get_dummies(data=df, drop_first=True, dtype=int)
df
```

Teraz, gdy dane są wstępnie przetworzone, możemy podzielić ramkę danych `df` na zmienne cech (X) i zmienną docelową (y). Zakładamy, że kolumna `bought` jest zmienną docelową.

```python
# Oddziel zmienne cech (X) od zmiennej docelowej (y)
X = df.drop('bought', axis=1)
y = df['bought']

print("Features (X) shape:", X.shape)
print("Target (y) shape:", y.shape)

# Wyświetl pierwsze 5 wierszy cech
print("\nFirst 5 rows of X:")
display(X.head())

# Wyświetl pierwsze 5 wartości zmiennej docelowej
print("\nFirst 5 values of y:")
display(y.head())
```
